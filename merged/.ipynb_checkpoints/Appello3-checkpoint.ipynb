```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import math
from pandas.tseries.holiday import USFederalHolidayCalendar
from pandas.tseries.offsets import CustomBusinessDay
from sklearn.preprocessing import LabelEncoder
from sklearn import preprocessing
import xgboost

from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
```

```python
train = pd.read_csv("data/train_users_2.csv")
sessions = pd.read_csv("data/sessions.csv")
```

```python
print("TRAIN columns:", train.columns.tolist())
print("\nSESSIONS columns:", sessions.columns.tolist())

display(train.dtypes.to_frame("dtype"))
display(sessions.dtypes.to_frame("dtype"))
```

```python
def missing_report(df, name):
    miss = df.isna().sum()
    miss = miss[miss > 0].sort_values(ascending=False)
    print(f"\n{name} - Missing columns: {len(miss)} / {df.shape[1]}")
    if len(miss):
        display(pd.DataFrame({
            "missing": miss,
            "missing_%": (miss / len(df) * 100).round(2)
        }))

missing_report(train, "train")
missing_report(sessions, "sessions")
```

```python
display(train.describe(include=[np.number]).T)
display(sessions.describe(include=[np.number]).T)
```

```python
if "id" in train.columns:
    print("train.id unique:", train["id"].nunique(), " / rows:", len(train))
if "user_id" in sessions.columns:
    print("sessions.user_id unique:", sessions["user_id"].nunique(), " / rows:", len(sessions))

# how many train users have sessions?
if "id" in train.columns and "user_id" in sessions.columns:
    overlap = sessions["user_id"].isin(train["id"]).mean()
    print("Share of sessions rows whose user_id exists in train.id:", round(overlap, 4))
```

```python
# =========================
# 1) Convert date columns properly
# =========================

train["date_account_created"] = pd.to_datetime(train["date_account_created"])
train["date_first_booking"] = pd.to_datetime(train["date_first_booking"], errors="coerce")

# timestamp_first_active conversion
train["timestamp_first_active"] = pd.to_datetime(
    train["timestamp_first_active"].astype(str),
    format="%Y%m%d%H%M%S"
)

print("Date conversion done.")
```

```python
# =========================
# 2) Fix age column professionally
# =========================

# Replace impossible ages
train.loc[train["age"] < 15, "age"] = np.nan
train.loc[train["age"] > 100, "age"] = np.nan

# Also handle birth year case (1900-2014 range)
train.loc[(train["age"] > 1900) & (train["age"] < 2015), "age"] = \
    2015 - train.loc[(train["age"] > 1900) & (train["age"] < 2015), "age"]

print("Age cleaned.")
```

```python
# =========================
# 3) Extract powerful date features
# =========================

train["dac_year"] = train["date_account_created"].dt.year
train["dac_month"] = train["date_account_created"].dt.month
train["dac_day"] = train["date_account_created"].dt.day
train["dac_weekday"] = train["date_account_created"].dt.weekday

train["tfa_year"] = train["timestamp_first_active"].dt.year
train["tfa_month"] = train["timestamp_first_active"].dt.month
train["tfa_day"] = train["timestamp_first_active"].dt.day

print("Date features created.")
```

```python
# =========================
# 4) Target Distribution (VERY IMPORTANT)
# =========================

plt.figure(figsize=(10,5))
train["country_destination"].value_counts().plot(kind="bar")
plt.title("Target Distribution")
plt.show()

train["country_destination"].value_counts(normalize=True)
```

```python
train.age.describe()
```

```python
# =========================
# SESSION FEATURE ENGINEERING - V1
# =========================

print("Starting session aggregation...")

# Fill missing categorical values first
sessions["action"] = sessions["action"].fillna("unknown")
sessions["action_type"] = sessions["action_type"].fillna("unknown")
sessions["action_detail"] = sessions["action_detail"].fillna("unknown")
sessions["device_type"] = sessions["device_type"].fillna("unknown")
sessions["secs_elapsed"] = sessions["secs_elapsed"].fillna(0)

# Basic aggregations
session_agg = sessions.groupby("user_id").agg(
    total_actions=("action", "count"),
    unique_actions=("action", "nunique"),
    unique_action_types=("action_type", "nunique"),
    unique_devices=("device_type", "nunique"),
    total_secs=("secs_elapsed", "sum"),
    mean_secs=("secs_elapsed", "mean"),
    median_secs=("secs_elapsed", "median"),
    std_secs=("secs_elapsed", "std")
).reset_index()

print("Aggregation complete.")
print(session_agg.shape)
session_agg.head()

```

```python
plt.figure(figsize=(8,5))
session_agg["total_actions"].hist(bins=50)
plt.title("Distribution of Total Actions per User")
plt.show()
```

```python
session_agg.describe()
```

```python
# =========================
# Log Transform Heavy-Tailed Features
# =========================

log_cols = [
    "total_actions",
    "total_secs",
    "mean_secs",
    "median_secs",
    "std_secs"
]

for col in log_cols:
    session_agg[f"log_{col}"] = np.log1p(session_agg[col])

print("Log transformation done.")
```

```python
plt.figure(figsize=(8,5))
session_agg["log_total_actions"].hist(bins=50)
plt.title("Log Distribution of Total Actions")
plt.show()
```

```python
# =========================
# Merge Train with Session Features
# =========================

train_merged = train.merge(
    session_agg,
    how="left",
    left_on="id",
    right_on="user_id"
)

print(train_merged.shape)
```

```python
# Check how many users have no sessions
no_session_ratio = train_merged["total_actions"].isna().mean()
print("Users without sessions ratio:", round(no_session_ratio, 4))
```
